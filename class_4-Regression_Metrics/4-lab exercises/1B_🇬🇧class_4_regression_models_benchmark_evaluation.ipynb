## Academic Introduction to the Study of Regression Metrics

This laboratory notebook primarily aims at **exploring and deeply understanding evaluation metrics for regression models**. Through practical examples and application exercises using a football player dataset, the goal is to consolidate knowledge on how to quantify and interpret the performance of predictive algorithms that estimate continuous values.

### What is Regression and its Metrics?

Regression is a fundamental technique in Machine Learning and statistics that aims to model the relationship between a continuous dependent variable (target) and one or more independent variables (features). Unlike classification, which predicts discrete categories, regression focuses on predicting numerical values, such as a player's market value, a house's price, or tomorrow's temperature.

The evaluation of regression models is crucial to determine their effectiveness and generalization. Metrics such as **Mean Absolute Error (MAE)**, **Mean Squared Error (MSE)**, **Root Mean Squared Error (RMSE)**, **Coefficient of Determination (R²)**, and **Mean Absolute Percentage Error (MAPE)** are discussed. Each offers a distinct perspective on model performance, whether in terms of error magnitude (MAE, RMSE), penalization of larger errors (MSE, RMSE), or the proportion of explained variance (R²).

### Why is this Study Important?

The correct selection and interpretation of regression metrics are vital for several reasons:

1.  **Rigorous Evaluation:** They allow objectively quantifying how well a model is predicting results on unseen data, avoiding overestimation or underestimation of its capability.
2.  **Decision Making:** Guides the choice of the best model among multiple alternatives, informing business or research decisions based on solid performance criteria.
3.  **Problem Identification:** Helps diagnose issues such as *overfitting* (when the model fits the training data too closely, losing generalization ability) or *underfitting* (when the model is too simple to capture data complexity).
4.  **Effective Communication:** Facilitates communication of model performance to technical and non-technical stakeholders, setting realistic and well-founded expectations.

### Use Cases

The applications of regression models and the need for their metrics are vast across various domains:

*   **Finance:** Predicting stock prices, credit risk assessment, market volatility estimation.
*   **Retail and Sales:** Projecting product sales, price optimization, demand forecasting.
*   **Healthcare:** Predicting hospital stay duration, optimal drug dosage, estimating treatment costs.
*   **Engineering and Manufacturing:** Predicting equipment failures, process optimization, energy consumption estimation.
*   **Sports:** Athlete performance evaluation (like player market value in this lab), game outcome prediction, training strategy optimization.

This laboratory provides the conceptual and practical foundation for students and professionals to confidently apply and interpret regression metrics, essential for developing and deploying robust and effective predictive models.

### Ethical, Fairness, and Governance Considerations in Regression

Beyond technical accuracy, the construction and deployment of regression models require deep reflection on their ethical and social implications, as well as adherence to governance principles and regulations. Ignoring these aspects can lead to harmful outcomes, discrimination, and loss of trust.

*   **Bias in Regression Models:** Regression models learn patterns from historical data. If this data contains implicit or explicit biases (e.g., over/under-representation of demographic groups, historical prejudices), the model can perpetuate or amplify these biases. It is crucial to assess whether the model performs unequally for different subgroups of the population (e.g., by gender, ethnicity, age, nationality), which can lead to unfair or discriminatory results. Fairness tools can help detect and mitigate these biases.

*   **Interpretability (Explainable AI - XAI):** In many applications, it is not enough for the model to make an accurate prediction; it is necessary to understand *why* it made that prediction. XAI tools (such as SHAP and LIME) help demystify the "black box" of models, revealing which features are most influential in predictions. This is crucial for ensuring trust, transparency, and fairness, especially in high-impact applications where decisions have significant consequences for individuals.

*   **Ethical Responsibility and Governance:** The development and deployment of AI systems must be guided by a sense of responsibility, considering individual and social impact. AI governance involves creating and applying clear guidelines for the model lifecycle, including regular audits, transparent documentation, and peer review processes. The goal is to ensure compliance with ethical standards, prevent unintentional harm, and promote the beneficial use of AI. Ethical responsibility goes beyond technical correctness, encompassing the evaluation of social consequences and the mitigation of potential harms.

*   **LGPD and Data Privacy:** In contexts where personal data is used to train and operate regression models, it is imperative to ensure compliance with the General Data Protection Law (LGPD) in Brazil (and other regulations like GDPR in Europe). This includes principles such as data minimization, anonymization/pseudonymization, obtaining explicit consent, guaranteeing data subject rights (access, correction, deletion), and implementing robust information security measures. Privacy must be incorporated from the system's design (*Privacy by Design*).

*   **Responsible AI Frameworks:** Organizations and governments have developed Responsible AI (RAI) principles and frameworks to guide the ethical development and use of artificial intelligence. These frameworks are generally based on pillars such as safety, fairness, explainability, robustness, privacy, and transparency. Familiarizing oneself with these principles helps incorporate ethics into the Machine Learning engineering process.

*   **Regulations like the AI Act:** Regulations are rapidly evolving to address the challenges of AI. The European Union's AI Act is a pioneering example, establishing legal and safety requirements for artificial intelligence systems, especially those considered high-risk. Understanding and monitoring these regulations is crucial for legal compliance and for building AI systems that are ethical, safe, and socially acceptable.

## 8. R² — Coefficient of Determination (manual and ready-made)

**Example 8.** R² = 1 − SSres/SStot: SSres = sum of squared model errors; SStot = sum of squared deviations from the mean.

```python
yv = np.array([2,5,6,8,9.]); yh = np.array([2.6,4.3,6,7.7,9.4])
SSres = ((yv-yh)**2).sum(); SStot = ((yv-yv.mean())**2).sum()
print('R² manual =', round(1-SSres/SStot,4), '| r2_score =', round(r2_score(yv,yh),4))
```

```python
r2_sk = r2_score(y_test, pred)
print('R² with r2_score:', round(r2_sk, 3))

SSres = ((y_test - pred)**2).sum()
SStot = ((y_test - y_test.mean())**2).sum()
r2_manual = 1 - (SSres / SStot)
print('R² manual:', round(r2_manual, 3))
```

## 9. R² = Pearson's r² (the class doubt)

**Example 9.** In SIMPLE linear regression (1 predictor), the model's R² is EXACTLY the square of Pearson's r. And a bad model gives a negative R², while Pearson's r between the variables doesn't change.

```python
Xe = np.array([1,2,3,4,5.]); Ye = np.array([2,5,6,8,9.])
r = np.corrcoef(Xe, Ye)[0,1]
reg = LinearRegression().fit(Xe.reshape(-1,1), Ye)
print('Pearson r =', round(r,4), '| r² =', round(r**2,4),
      '| Regression R² =', round(r2_score(Ye, reg.predict(Xe.reshape(-1,1))),4))
print('Bad model (always predicts 3): R² =', round(r2_score(Ye, np.full(5,3.0)),3),
      '| Pearson remains', round(r,3))
```

**Exercise 9 (application).** In the dataset, use only the `indice_popularidade` (popularity index) feature to predict the target: calculate the **Pearson's r** between them, the **r²**, and the **R²** of a simple linear regression with this feature. Do they match?

```python
X_simple = df[['indice_popularidade']]
y_simple = df[ALVO]

r_pearson = X_simple['indice_popularidade'].corr(y_simple)
r2_pearson = r_pearson**2
print('Pearson r:', round(r_pearson, 3))
print('r² of Pearson:', round(r2_pearson, 3))

reg_simple = LinearRegression().fit(X_simple, y_simple)
r2_reg_simple = r2_score(y_simple, reg_simple.predict(X_simple))
print('R² of simple linear regression:', round(r2_reg_simple, 3))

print('\nYes, they match! In simple linear regression (with 1 predictor), the model\'s R² is the square of Pearson\'s correlation coefficient.')
```

## 23. Feature Importance for Model Fairness and Transparency

Understanding which characteristics (features) are most influential in a model's predictions is fundamental to assessing its fairness and ensuring transparency. In linear models, coefficients can indicate the direction and magnitude of each feature's importance. By visualizing and analyzing feature importance, we can identify potential biases or unexpected model dependencies.

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Retrain the LinearRegression model (or use the already trained 'pipe_lr' from Ex. 19 if applicable)
# Ensure pipe_lr is available and trained with X_train, y_train
# The pipe_lr already includes StandardScaler and LinearRegression
pipe_lr.fit(X_train, y_train)

# Access the coefficients of the LinearRegression model within the pipeline
# The last step of the pipeline is LinearRegression
linear_model = pipe_lr.named_steps['linearregression']

# Create a DataFrame for the coefficients
feature_importances = pd.DataFrame({
    'Feature': X_train.columns,
    'Importance': linear_model.coef_
})

# Sort by absolute magnitude of importance
feature_importances['Abs_Importance'] = feature_importances['Importance'].abs()
feature_importances = feature_importances.sort_values(by='Abs_Importance', ascending=False)

plt.figure(figsize=(10, 6))
sns.barplot(x='Importance', y='Feature', data=feature_importances)
plt.title('Feature Importance (Linear Model Coefficients)')
plt.xlabel('Coefficient (Importance)')
plt.ylabel('Feature')
plt.grid(axis='x', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```
