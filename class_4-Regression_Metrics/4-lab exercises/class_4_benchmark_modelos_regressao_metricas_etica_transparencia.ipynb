## Introdução Acadêmica ao Estudo de Métricas de Regressão

Este notebook de laboratório tem como objetivo principal a **exploração e compreensão aprofundada das métricas de avaliação para modelos de regressão**. Através de exemplos práticos e exercícios de aplicação utilizando um dataset de jogadores de futebol, busca-se consolidar o conhecimento sobre como quantificar e interpretar a performance de algoritmos preditivos que estimam valores contínuos.

### O que é Regressão e suas Métricas?

A regressão é uma técnica fundamental em *Machine Learning* e estatística que visa modelar a relação entre uma variável dependente (alvo) contínua e uma ou mais variáveis independentes (features). Diferentemente da classificação, que prevê categorias discretas, a regressão foca na previsão de valores numéricos, como o valor de mercado de um jogador, o preço de uma casa ou a temperatura de amanhã.

A avaliação de modelos de regressão é crucial para determinar sua eficácia e generalização. Métricas como o **Erro Absoluto Médio (MAE)**, **Erro Quadrático Médio (MSE)**, **Raiz do Erro Quadrático Médio (RMSE)**, **Coeficiente de Determinação (R²)** e **Erro Percentual Absoluto Médio (MAPE)** são discutidas. Cada uma oferece uma perspectiva distinta sobre o desempenho do modelo, seja em termos de magnitude do erro (MAE, RMSE), penalização de erros maiores (MSE, RMSE) ou proporção da variância explicada (R²).

### Por que este Estudo é Importante?

A seleção e a interpretação correta das métricas de regressão são vitais por diversas razões:

1.  **Avaliação Rigorosa:** Permitem quantificar objetivamente quão bem um modelo está prevendo os resultados em dados não vistos, evitando a superestimação ou subestimação de sua capacidade.
2.  **Tomada de Decisão:** Guia a escolha do melhor modelo entre múltiplas alternativas, informando decisões de negócios ou pesquisa com base em critérios de desempenho sólidos.
3.  **Identificação de Problemas:** Ajuda a diagnosticar questões como *overfitting* (quando o modelo se ajusta demais aos dados de treino, perdendo capacidade de generalização) ou *underfitting* (quando o modelo é muito simples para capturar a complexidade dos dados).
4.  **Comunicação Efetiva:** Facilita a comunicação do desempenho do modelo para stakeholders técnicos e não-técnicos, estabelecendo expectativas realistas e embasadas.

### Casos de Uso (Use Cases)

As aplicações de modelos de regressão e a necessidade de suas métricas são vastas em diversos domínios:

*   **Finanças:** Previsão de preços de ações, avaliação de risco de crédito, estimativa de volatilidade de mercado.
*   **Comércio e Vendas:** Projeção de vendas de produtos, otimização de preços, previsão de demanda.
*   **Saúde:** Predição de tempo de internação hospitalar, dosagem ideal de medicamentos, estimativa de custos de tratamento.
*   **Engenharia e Manufatura:** Previsão de falhas de equipamentos, otimização de processos, estimativa de consumo de energia.
*   **Esportes:** Avaliação de desempenho de atletas (como o valor de mercado de jogadores neste laboratório), previsão de resultados de jogos, otimização de estratégias de treino.

Este laboratório fornece a base conceitual e prática para que os estudantes e profissionais possam aplicar e interpretar métricas de regressão com confiança, fundamental para o desenvolvimento e implantação de modelos preditivos robustos e eficazes.

### Considerações Éticas, de Justiça e Governança na Regressão

Além da precisão técnica, a construção e implantação de modelos de regressão exigem uma profunda reflexão sobre suas implicações éticas e sociais, bem como a adesão a princípios de governança e regulamentações. Ignorar esses aspectos pode levar a resultados prejudiciais, discriminação e perda de confiança.

*   **Viés em Modelos de Regressão:** Modelos de regressão aprendem padrões a partir dos dados históricos. Se esses dados contêm vieses implícitos ou explícitos (ex: super/sub-representação de grupos demográficos, preconceitos históricos), o modelo pode perpetuar ou amplificar esses vieses. É crucial avaliar se o modelo performa de forma desigual para diferentes subgrupos da população (ex: por gênero, etnia, idade, nacionalidade), o que pode levar a resultados injustos ou discriminatórios. Ferramentas de *fairness* podem auxiliar na detecção e mitigação desses vieses.

*   **Interpretabilidade (Explainable AI - XAI):** Em muitas aplicações, não basta que o modelo faça uma previsão precisa; é necessário entender *por que* ele fez essa previsão. Ferramentas de XAI (como SHAP e LIME) ajudam a desmistificar a "caixa preta" dos modelos, revelando quais características são mais influentes nas previsões. Isso é crucial para garantir a confiança, a transparência e a justiça, especialmente em aplicações de alto impacto onde as decisões têm consequências significativas para indivíduos.

*   **Responsabilidade Ética e Governança:** O desenvolvimento e a implantação de sistemas de IA devem ser guiados por um senso de responsabilidade, considerando o impacto social e individual. A governança de IA envolve a criação e aplicação de diretrizes claras para o ciclo de vida dos modelos, incluindo auditorias regulares, documentação transparente e processos de revisão por pares. O objetivo é garantir a conformidade com padrões éticos, evitar danos não intencionais e promover o uso benéfico da IA. A responsabilidade ética vai além da correção técnica, abrangendo a avaliação das consequências sociais e a mitigação de danos potenciais.

*   **LGPD e Privacidade de Dados:** Em contextos onde dados pessoais são utilizados para treinar e operar modelos de regressão, é imperativo garantir a conformidade com a Lei Geral de Proteção de Dados (LGPD) no Brasil (e outras regulamentações como GDPR na Europa). Isso inclui princípios como minimização de dados, anonimização/pseudonimização, obtenção de consentimento explícito, garantia de direitos dos titulares dos dados (acesso, correção, exclusão) e implementação de medidas robustas de segurança da informação. A privacidade deve ser incorporada desde o design do sistema (*Privacy by Design*).

*   **Responsible AI Frameworks:** Organizações e governos têm desenvolvido princípios e frameworks de Responsible AI (IA Responsável) para guiar o desenvolvimento e uso ético da inteligência artificial. Esses frameworks geralmente se baseiam em pilares como segurança, justiça, explicabilidade, robustez, privacidade e transparência. Familiarizar-se com esses princípios ajuda a incorporar a ética no processo de engenharia de Machine Learning.

*   **Regulamentações como o AI Act:** As regulamentações estão evoluindo rapidamente para endereçar os desafios da IA. O AI Act da União Europeia é um exemplo pioneiro, estabelecendo requisitos legais e de segurança para sistemas de inteligência artificial, especialmente aqueles considerados de alto risco. Compreender e monitorar essas regulamentações é crucial para a conformidade legal e para a construção de sistemas de IA que sejam éticos, seguros e socialmente aceitáveis.

## 8. R² — coeficiente de determinação (na mão e pronto)

**Exemplo 8.** R² = 1 − SSres/SStot: SSres = soma dos erros² do modelo; SStot = soma dos desvios² da média.

```python
yv = np.array([2,5,6,8,9.]); yh = np.array([2.6,4.3,6,7.7,9.4])
SSres = ((yv-yh)**2).sum(); SStot = ((yv-yv.mean())**2).sum()
print('R² na mão =', round(1-SSres/SStot,4), '| r2_score =', round(r2_score(yv,yh),4))
```

**Exercício 8 (aplicação).** Calcule o **R²** do modelo no teste com `r2_score` e confirme na mão (1 − SSres/SStot).

```python
r2_sk = r2_score(y_test, pred)
print('R² com r2_score:', round(r2_sk, 3))

SSres = ((y_test - pred)**2).sum()
SStot = ((y_test - y_test.mean())**2).sum()
r2_manual = 1 - (SSres / SStot)
print('R² na mão:', round(r2_manual, 3))
```

## 9. R² = r² do Pearson (a dúvida da aula)

**Exemplo 9.** em regressão linear SIMPLES (1 preditor), o R² do modelo é EXATAMENTE o quadrado do Pearson. E um modelo ruim dá R² negativo, enquanto o Pearson entre as variáveis não muda.

```python
Xe = np.array([1,2,3,4,5.]); Ye = np.array([2,5,6,8,9.])
r = np.corrcoef(Xe, Ye)[0,1]
reg = LinearRegression().fit(Xe.reshape(-1,1), Ye)
print('Pearson r =', round(r,4), '| r² =', round(r**2,4),
      '| R² da regressão =', round(r2_score(Ye, reg.predict(Xe.reshape(-1,1))),4))
print('Modelo ruim (prevê sempre 3): R² =', round(r2_score(Ye, np.full(5,3.0)),3),
      '| Pearson continua', round(r,3))
```

**Exercício 9 (aplicação).** No dataset, use só a feature `indice_popularidade` para prever o alvo: calcule o **Pearson r** entre elas, o **r²**, e o **R²** de uma regressão linear simples com essa feature. Eles batem?

```python
X_simple = df[['indice_popularidade']]
y_simple = df[ALVO]

r_pearson = X_simple['indice_popularidade'].corr(y_simple)
r2_pearson = r_pearson**2
print('Pearson r:', round(r_pearson, 3))
print('r² do Pearson:', round(r2_pearson, 3))

reg_simple = LinearRegression().fit(X_simple, y_simple)
r2_reg_simple = r2_score(y_simple, reg_simple.predict(X_simple))
print('R² da regressão linear simples:', round(r2_reg_simple, 3))

print('\nSim, eles batem! Em regressão linear simples (com 1 preditor), o R² do modelo é o quadrado do coeficiente de correlação de Pearson.')
```

## 10. Baseline: prever sempre a média (R² = 0)

## 23. Importância das Features para Fairness e Transparência do Modelo

Compreender quais características (features) são mais influentes nas previsões de um modelo é fundamental para avaliar sua justiça (fairness) e garantir a transparência. Em modelos lineares, os coeficientes podem indicar a direção e a magnitude da importância de cada feature. Ao visualizar e analisar a importância das features, podemos identificar potenciais vieses ou dependências inesperadas do modelo.

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Re-treinar o modelo LinearRegression (ou usar o já treinado 'pipe_lr' do Ex. 19 se for o caso)
# Certificar que o pipe_lr está disponível e treinado com X_train, y_train
# O pipe_lr já inclui o StandardScaler e LinearRegression
pipe_lr.fit(X_train, y_train)

# Acessar os coeficientes do modelo LinearRegression dentro do pipeline
# O último passo do pipeline é a LinearRegression
linear_model = pipe_lr.named_steps['linearregression']

# Criar um DataFrame para os coeficientes
feature_importances = pd.DataFrame({
    'Feature': X_train.columns,
    'Importance': linear_model.coef_
})

# Classificar pela magnitude absoluta da importância
feature_importances['Abs_Importance'] = feature_importances['Importance'].abs()
feature_importances = feature_importances.sort_values(by='Abs_Importance', ascending=False)

plt.figure(figsize=(10, 6))
sns.barplot(x='Importance', y='Feature', data=feature_importances)
plt.title('Importância das Features (Coeficientes do Modelo Linear)')
plt.xlabel('Coeficiente (Importância)')
plt.ylabel('Feature')
plt.grid(axis='x', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```
