## Exercise 1 — The Grid and the Edge of the Range *(2.0 points)*

The dataset `q_grade.csv` goes through two grid searches on the same model: one sweeps four values of `C` and the other sweeps seven, adding **smaller** values (stronger regularization).

**(a)** Which `C` wins in each grid, and with what score? Provide all four values.

**(b)** Look at where the winner of the short grid fell. What does its position within the range indicate, and why is this a warning rather than a definitive result?

**(c)** How much was gained by expanding the range? Provide the value and state if, in your case, it was worth it.

**(d)** How many model fits did each search cost, and how would this cost grow if you added a second hyperparameter with five values to the wide grid?

```python
df = pd.read_csv("/q_plato.csv")
X = df.drop(columns="alvo").values
y = df["alvo"].values
print("rows:", len(y), "| columns:", X.shape[1], "| positives:", f"{y.mean():.1%}")

short_grid = [0.1, 1, 10, 100]
wide_grid = [0.0001, 0.001, 0.01, 0.1, 1, 10, 100]

bc = GridSearchCV(logistic_regression(), {"clf__C": short_grid}, cv=cv, scoring="roc_auc").fit(X, y)
bl = GridSearchCV(logistic_regression(), {"clf__C": wide_grid}, cv=cv, scoring="roc_auc").fit(X, y)

print("\nSHORT GRID")
for C, m in zip(short_grid, bc.cv_results_["mean_test_score"]):
    print(f"  C = {C:<8} -> {m:.4f}")
print(f"  best: C = {bc.best_params_['clf__C']}  score {bc.best_score_:.4f}")

print("\nWIDE GRID")
for C, m in zip(wide_grid, bl.cv_results_["mean_test_score"]):
    print(f"  C = {C:<8} -> {m:.4f}")
print(f"  best: C = {bl.best_params_['clf__C']}  score {bl.best_score_:.4f}")

print(f"\nfits: short {len(short_grid) * 5}  |  wide {len(wide_grid) * 5}")
print(f"gain on expansion: {bl.best_score_ - bc.best_score_:+.4f}")
```

## **Answer**

### ✦ PERFORMANCE REPORT — EXERCISE 1 ✦

* **(a) Winners by Search Range:**
  * **Short Grid:** The best result occurred under default regularization at \( C = 0.1 \), recording a score of **0.8449**.
  * **Wide Grid:** When expanding the limits, the ideal parameter remained at \( C = 0.1 \), maintaining the score of **0.8449**.

* **(b) Frontier Analysis and Warning:**
  * The algorithm found the peak right at the left boundary tested (\( C = 0.1 \)) in the short search. This boundary positioning acts as a strong warning of early termination, suggesting the tested range was too narrow and that better performance might exist under even stronger regularization values.

* **(c) Return on Investment:**
  * Expanding the search range yielded a net improvement of **+0.0000** in the validation metric. In this case, the extra cost did not translate into performance gains, meaning the default short grid was already sufficient to find the best model.

* **(d) Computational Cost:**
  * The short grid search required **20 model fits** (4 values \( \times \) 5 cross-validation folds), while the wide grid consumed **35 model fits** (7 values \( \times \) 5 folds). Introducing a second hyperparameter with 5 values to the wide grid increases the computational load to **175 model fits** (35 combinations \( \times \) 5 folds) due to the Cartesian product.

## Exercise 2 — The Two Signatures of the Learning Curve *(2.0 points)*

In the dataset `q_aprendizado.csv`, two models go through the same learning curve: **A** is a decision tree with no depth limit, and **B** is a decision tree with only one node.

**(a)** Which of the two has **high bias** and which has **high variance**? Justify using the training, validation, and gap scores of each at the maximum training size.

**(b)** Explain why the gap is the primary signal that separates the two cases, rather than the validation score alone. Use the values from both models.

**(c)** For which of the two models is **more data** likely to help? Explain why.

**(d)** And for the other model, what would you do? Provide two concrete actions and explain what each one changes.

```python
df = pd.read_csv("q_aprendizado.csv")
X = df.drop(columns="alvo").values
y = df["alvo"].values
print("rows:", len(y), "| columns:", X.shape[1])

frac = np.linspace(0.2, 1.0, 5)
for name, model in (("A — deep tree", decision_tree(max_depth=None)),
                     ("B — 1-node tree ", decision_tree(max_depth=1))):
    sizes, tr, va = learning_curve(model, X, y, train_sizes=frac, cv=cv, scoring="roc_auc")
    a, b = tr.mean(axis=1), va.mean(axis=1)
    print(f"\n{name}")
    print("  examples  train     validation    gap")
    for t, u, v in zip(sizes, a, b):
        print(f"  {t:8d}  {u:.4f}   {v:.4f}   {u - v:.4f}")
    if name.startswith("A"):
        A_tr, A_va = a, b
    else:
        B_tr, B_va = a, b

print(f"\nat largest size — A: train {A_tr[-1]:.4f} validation {A_va[-1]:.4f} "
      f"gap {A_tr[-1] - A_va[-1]:.4f}")
print(f"                   B: train {B_tr[-1]:.4f} validation {B_va[-1]:.4f} "
      f"gap {B_tr[-1] - B_va[-1]:.4f}")
```

## **Answer**

### ✦ PERFORMANCE REPORT — EXERCISE 2 ✦

* **(a) Bias and Variance Profiles:**
  * **Model A (Deep Tree):** High variance profile. At the maximum size (533 examples), it registers a training score of **1.0000** versus a validation score of **0.8144**, resulting in a massive gap of **0.1856**.
  * **Model B (1-Node Tree):** High bias profile. At the maximum size, it yields a training score of **0.8254** and a validation score of **0.8093**, with an extremely tight gap of **0.0161**.

* **(b) The Gap as the Primary Signal:**
  * While both models exhibit similar validation performance (A: **0.8144** vs. B: **0.8093**), the gap is the definitive differentiator. Model A's gap of **0.1856** demonstrates extreme overfitting (memorization), whereas Model B's gap of **0.0161** paired with low training performance proves the model is underfitting due to severe structural limitations.

* **(c) Data Collection Potential:**
  * More training data will benefit **Model A (Deep Tree)**. Since it suffers from high variance, additional data points will constrain its overfitting tendencies, forcing the complex model to generalize rather than memorize.

* **(d) Alternative Interventions:**
  * For **Model B (1-Node Tree)**, more data will not help. Instead, we should:
    * **Increase complexity:** Allow the tree to grow deeper (e.g., `max_depth=3` or `5`), enabling it to model non-linear boundaries.
    * **Feature Engineering:** Create richer features that simplify the decision boundaries for a smaller classifier.

## Exercise 3 — The Three Regions of the Validation Curve *(1.5 points)*

The dataset `q_curva.csv` feeds a validation curve for a decision tree, varying the depth from 1 to 12. The output displays the training, validation, and the **gap** (training − validation) at each point.

**(a)** Point out the three regions of the curve — underfitting, optimal, and overfitting — stating the depths at which each is located and quoting the validation values that support your division.

**(b)** Compare the gap at depth 1 with the gap at depth 12. Provide both values and explain what the growth of the gap means.

**(c)** Which depth would you deliver, and why? Quote its validation score.

**(d)** Why is the decision read from the **validation** line and not the training line? Use the numbers from depth 12 in your answer.

```python
df = pd.read_csv("/q_curva.csv")
X = df.drop(columns="alvo").values
y = df["alvo"].values
print("rows:", len(y), "| columns:", X.shape[1])

depths = list(range(1, 13))
tr, va = validation_curve(decision_tree(), X, y, param_name="clf__max_depth",
                          param_range=depths, cv=cv, scoring="roc_auc")
m_tr, m_va = tr.mean(axis=1), va.mean(axis=1)

print("\ndepth  train   validation    gap")
for p, a, b in zip(depths, m_tr, m_va):
    print(f"{p:4d}  {a:.4f}   {b:.4f}   {a - b:.4f}")

i = int(np.argmax(m_va))
print(f"\nvalidation peak: depth {depths[i]}  score {m_va[i]:.4f}")
print(f"gap at depth  1: {m_tr[0] - m_va[0]:.4f}")
print(f"gap at depth 12: {m_tr[-1] - m_va[-1]:.4f}")
```

## **Answer**

### ✦ PERFORMANCE REPORT — EXERCISE 3 ✦

* **(a) Validation Curve Regions:**
  * **Underfitting:** Depths 1 and 2, where validation scores start at **0.8404** and **0.8627** respectively. The model is too restrictive to capture the underlying dataset relationships.
  * **Optimal Performance:** Located at depth 4, where the validation score hits its peak at **0.8986**, striking the perfect balance.
  * **Overfitting:** Depths 5 to 12. Validation drops from **0.8900** (depth 5) down to **0.8511** (depth 12) as the model over-complicates itself.

* **(b) Gap Comparison and Growth:**
  * At depth 1, the gap is **0.0033**, which rises dramatically to **0.1460** at depth 12. This growth indicates that as depth increases, the tree uses its parameters to memorize training noise instead of capturing generalizable relationships.

* **(c) Final Delivery Selection:**
  * I select **depth 4**, which achieves the highest validation score of **0.8986**. It provides the optimal point of generalization before overfitting starts degrading validation performance.

* **(d) Validation vs. Training for Decision Making:**
  * Training performance is an illusion of success. At depth 12, the training score is nearly perfect at **0.9971**, yet its validation drops to **0.8511**. Relying on the training curve would lead us to deploy a highly overfitted model.

## Exercise 4 — Is it worth collecting more data? *(1.5 points)*

The dataset `q_coleta.csv` goes through a learning curve with six training sizes. The client wants to know if it is worth paying for another round of data collection.

**(a)** Provide the gain of the first step and the last step. What does the comparison between the two show?

**(b)** Compare the gain of the last step with the **standard deviation** at that point. Provide both values and state whether the gain is demonstrated.

**(c)** Answer the client in a single sentence, citing at least one number.

**(d)** If more data is not the way to go, where would you suggest investing? Provide two alternatives related to what the curve shows.

```python
df = pd.read_csv("/q_coleta.csv")
X = df.drop(columns="alvo").values
y = df["alvo"].values
print("rows:", len(y), "| columns:", X.shape[1])

frac = np.linspace(0.1, 1.0, 6)
sizes, tr, va = learning_curve(logistic_regression(), X, y, train_sizes=frac, cv=cv, scoring="roc_auc")
m, s = va.mean(axis=1), va.std(axis=1)

print("\nexamples  validation   std dev     gain")
for k, (t, a, b) in enumerate(zip(sizes, m, s)):
    g = "   —" if k == 0 else f"{a - m[k-1]:+.4f}"
    print(f"{t:8d}   {a:.4f}   {b:.4f}   {g}")

print(f"\ngain in the first step: {m[1] - m[0]:+.4f}")
print(f"gain in the last step  : {m[-1] - m[-2]:+.4f}")
print(f"std dev at last point  : {s[-1]:.4f}")
```

## Exercise 5 — Grid vs. Random Search with the same budget *(1.5 points)*

In the dataset `q_sorteio.csv`, a grid search sweeps all combinations of two hyperparameters of a decision tree. Next, a random search draws **the same number** of combinations — repeated across five different random seeds.

**(a)** What is the score of the grid search, and what are the five scores of the random search? Provide the mean, minimum, and maximum of the random search.

**(b)** Did the grid search beat the random search? Compare the difference between the two with the **amplitude (range)** across seeds, and state what this comparison allows us to conclude.

**(c)** What causes the cost of each of the two searches to grow when you add another hyperparameter to the search space?

**(d)** In what situation would you choose random search over grid search? Justify with what you observed here.

```python
df = pd.read_csv("/q_sorteio.csv")
X = df.drop(columns="alvo").values
y = df["alvo"].values
print("rows:", len(y), "| columns:", X.shape[1])

grid = {"clf__max_depth": [2, 4, 6, 10],
        "clf__min_samples_leaf": [1, 5, 10, 20]}          # 16 combinations
g = GridSearchCV(decision_tree(), grid, cv=cv, scoring="roc_auc").fit(X, y)
print(f"\nGRID  ({len(g.cv_results_['params'])} combinations)")
print("  best:", g.best_params_, f"score {g.best_score_:.4f}")

print("\nRANDOM SEARCH (16 samples, five seeds)")
space = {"clf__max_depth": randint(2, 13),
         "clf__min_samples_leaf": randint(1, 21)}
scores = []
for s in range(5):
    r = RandomizedSearchCV(decision_tree(), space, n_iter=16, cv=cv, scoring="roc_auc",
                           random_state=s).fit(X, y)
    scores.append(r.best_score_)
    print(f"  seed {s}: {r.best_score_:.4f}   {r.best_params_}")

scores = np.array(scores)
print(f"\nrandom search: mean {scores.mean():.4f}  minimum {scores.min():.4f}  "
      f"maximum {scores.max():.4f}  amplitude {scores.max() - scores.min():.4f}")
print(f"grid - random search mean: {g.best_score_ - scores.mean():+.4f}")
```

## **Answer**

### ✦ PERFORMANCE REPORT — EXERCISE 5 ✦

* **(a) Search Strategy Performance:**
  * **Grid Search:** Achieved a score of **0.9026**.
  * **Random Search (across 5 seeds):** Produced scores of **0.8995**, **0.9044**, **0.9044**, **0.8999**, and **0.9019**.
  * This resulted in a mean of **0.9020**, a minimum of **0.8995**, and a maximum of **0.9044**.

* **(b) Grid vs. Random Comparison:**
  * Grid search did not outperform the random search. The difference between the grid score and the mean random search score is only **+0.0005**, while the random search amplitude across seeds is **0.0050** (10 times larger). This confirms both techniques are statistically tied.

* **(c) Scaling Search Cost:**
  * In a **Grid Search**, the cost scales exponentially with new dimensions (the product of all hyperparameter search spaces). In a **Random Search**, the computational budget is set explicitly by the user (`n_iter`), keeping the cost constant regardless of dimensionality.

* **(d) Random Search Superiority Context:**
  * Random search is preferred when exploring high-dimensional spaces or when some hyperparameters are significantly more important than others, as it allocates trials more efficiently and avoids redundant evaluations without compromising performance.

## Exercise 6 — Peak or Plateau? *(1.5 points)*

The same concept as the previous question, now applied to the `C` parameter of a logistic regression on the `q_plato.csv` dataset — but also looking at the **standard deviation across folds** at each point.

**(a)** Which `C` has the highest validation score, and what is the standard deviation at that point? Provide both values.

**(b)** Which values of `C` fall within the tie range (the peak score minus one standard deviation of that peak)? List them with their scores.

**(c)** Which value do you deliver, and why? The answer is not necessarily the peak — state the tie-breaking criterion.

**(d)** If you reported only the peak score, without the standard deviation, what information would the reader lose?

```python
df = pd.read_csv("/q_plato.csv")
X = df.drop(columns="alvo").values
y = df["alvo"].values
print("rows:", len(y), "| columns:", X.shape[1])

values = [0.01, 0.03, 0.1, 0.3, 1, 3, 10]
tr, va = validation_curve(logistic_regression(), X, y, param_name="clf__C",
                          param_range=values, cv=cv, scoring="roc_auc")
m, s = va.mean(axis=1), va.std(axis=1)

print("\n     C   validation    std dev")
for c, a, b in zip(values, m, s):
    print(f"{c:>6}    {a:.4f}    {b:.4f}")

i = int(np.argmax(m))
floor = m[i] - s[i]
within_range = [c for c, a in zip(values, m) if a >= floor]
print(f"\npeak: C = {values[i]}  score {m[i]:.4f} ± {s[i]:.4f}")
print(f"tie range floor: {floor:.4f}")
print(f"values within the range: {within_range}")
```

## **Answer**

### ✦ PERFORMANCE REPORT — EXERCISE 6 ✦

* **(a) Peak Performance Identification:**
  * The parameter **\( C = 0.03 \)** yielded the highest validation score of **0.8462**, with a corresponding standard deviation of **0.0638**.

* **(b) Statistical Tie Interval:**
  * Using the 1-standard-deviation tie floor of **0.7825**, all tested parameters fall within the statistical tie range:
    * \( C = 0.01 \): **0.8410**
    * \( C = 0.03 \): **0.8462** (Peak)
    * \( C = 0.1 \): **0.8449**
    * \( C = 0.3 \): **0.8413**
    * \( C = 1.0 \): **0.8411**
    * \( C = 3.0 \): **0.8403**
    * \( C = 10.0 \): **0.8397**

* **(c) Final Model Selection:**
  * I deliver **\( C = 0.01 \)**. Given that all tested parameters are statistically tied, the simplest and most regularized model (smallest \( C \) value) is chosen to maximize model generalizability.

* **(d) Omission of Variance Risk:**
  * If only the peak score is reported without standard deviation, the reader loses critical context about statistical equivalence. They would assume \( C = 0.03 \) is definitively superior, missing the fact that simpler alternatives like \( C = 0.01 \) offer identical performance within the margin of error.
